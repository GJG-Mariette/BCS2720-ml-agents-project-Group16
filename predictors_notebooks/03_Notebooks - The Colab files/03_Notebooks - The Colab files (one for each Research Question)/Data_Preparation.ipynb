```python
import pandas as pd
import glob
import os


RAW_PATH = '/content/drive/MyDrive/RL_Predictor_Project/01_Data_Raw'
PROCESSED_PATH = '/content/drive/MyDrive/RL_Predictor_Project/02_Data_Processed'


INPUT_FEATURES = [
    'enviroment', 'algorithm', 'learning_rate', 'batch_size', 'hidden_units',
    'num_layers', 'max_steps', 'ram_gb', 'cpu_cores', 'time_horizon',
    'buffer_size', 'beta', 'epsilon', 'lambd', 'num_epoch', 'tau',
    'num_parallel_agents', 'gamma', 'strength', 'gpu_available',
    'gpu_memory_gb', 'os'
]

TARGETS = [
    'training_duration_seconds',
    'final_mean_reward',
    'peak_ram_mb'
]


all_files = glob.glob(os.path.join(RAW_PATH, "*.csv"))
df_list = []

for f in all_files:
    temp_df = pd.read_csv(f)

    valid_cols = [c for c in (INPUT_FEATURES + TARGETS) if c in temp_df.columns]
    df_list.append(temp_df[valid_cols])

master_df = pd.concat(df_list, ignore_index=True)

if 'enviroment' in master_df.columns:
    master_df = master_df.rename(columns={'enviroment': 'environment'})


master_df['gpu_available'] = master_df['gpu_available'].fillna(False)
master_df['gpu_memory_gb'] = master_df['gpu_memory_gb'].fillna(0)

# 5. Save the Master File
os.makedirs(PROCESSED_PATH, exist_ok=True)
master_df.to_csv(os.path.join(PROCESSED_PATH, "master_data.csv"), index=False)

print(f"Data Preparation Complete! Saved {len(master_df)} rows to master_data.csv")
```
